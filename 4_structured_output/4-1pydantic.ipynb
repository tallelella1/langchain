##### 常见有 *Pydantic ；TypeDict ；JSON Schema格式 ；@dataclass*，但只有Pydanntic会校验格式，例如在在常见一个伪模型时用 year1 name2来表示年份和名字的字段，在输入时用year和name，只有Pytantic会报错，其余3种结构化输出不会校验字段是否相同

### 如果需要显示更多的模型输出信息，如token消耗 使用 include_raw=True 例如 structured_llm = model.with_structured_output(#Person #自己写的结构化的内容#，include_raw = True)
输出包含了完整的输出响应，包含三个字段
raw：返回的原始AIMessage。
parsed：解析后的输出
parsing_error：解析错误，当前用的是Pydantic，校验，格式不符合schema会导致报错。其它
三种方式不符合schema不会导致报错。
