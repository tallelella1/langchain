## Provider-agnostic middleware（与提供商无关的中间件）

The following middleware work with any LLM provider:  
以下中间件可与任何 LLM 提供商配合使用：

| Middleware（中间件） | Description（描述） |
|---|---|
| [Tool error（工具错误）] Catch tool execution exceptions and convert them to error messages for the model.|（捕获工具执行异常，并将其转换为发送给模型的错误消息。） 
| [Tool retry（工具重试）] Automatically retry failed tool calls with exponential backoff.|（使用指数退避自动重试失败的工具调用。） 
| [Model retry（模型重试）] Automatically retry failed model calls with exponential backoff.|（使用指数退避自动重试失败的模型调用。） 
| [Model fallback（模型回退）] Automatically fall back to alternative models when the primary model fails.|（主模型失败时，自动切换到备用模型。） 
| [Summarization（总结）] Summarize conversation history as it approaches token limits.|（会话历史接近令牌限制时自动进行总结。） 
| [Human-in-the-loop（人机交互）] Pause execution for human approval of tool calls.|（暂停执行，等待人工批准工具调用。） 
| [Model call limit（模型调用限制）] Limit model calls to help prevent excessive costs.|（限制模型调用次数，以控制成本。） 
| [Tool call limit（工具调用限制）] Control tool execution by limiting call counts.|（通过限制调用次数来控制工具执行。） 
| [PII detection（PII 检测）] Detect and handle personally identifiable information (PII).|（检测并处理个人身份信息。） 
| [To-do list（待办事项清单）] Give agents task-planning and tracking capabilities.|（为代理提供任务规划与跟踪能力。） 
| [LLM tool selector（LLM 工具选择器）] Use an LLM to select relevant tools before calling the main model.|（调用主模型前，使用 LLM 选择相关工具。） 
| [Provider tool search（提供商工具搜索）] Defer tools to provider-side search and surface them on demand.|（将工具交由提供商侧搜索，并按需提供。） 
| [Shell tool（Shell 工具）] Expose a persistent shell session for command execution.|（提供持久化 Shell 会话以执行命令。） 
| [Filesystem（文件系统）] Provide a filesystem for storing context and long-term memories.|（提供文件系统，用于存储上下文和长期记忆。） 
| [Subagent（子代理）] Add the ability to spawn subagents.|（支持创建子代理。） 
| [Rubric grading (Beta)（评分标准，测试版）] Use LLM-as-a-judge grading so agents can evaluate and iterate until they meet a rubric.|（使用 LLM 作为评判者，让代理进行评估和迭代，直到符合评分标准。） |
| [File search（文件搜索）] Provide Glob and Grep search tools for filesystem files.|（提供用于搜索文件系统的 Glob 和 Grep 工具。） 
| [Context editing（上下文编辑）] Manage conversation context by trimming or clearing tool uses.|（通过精简或清除工具调用内容来管理会话上下文。） 
| [LLM tool emulator（LLM 工具模拟器）] Emulate tool execution with an LLM for testing.|（使用 LLM 模拟工具执行，以便进行测试。）
