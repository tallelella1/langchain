## invoke

### 1 invoke传参

#### 1-1 文本输入

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

load_dotenv(override=True)

BAILIAN_API_KEY = os.getenv("BAILIAN_API_KEY")
BAILIAN_API_URL = os.getenv("BAILIAN_API_URL")

model = init_chat_model(
    model = "openai:qwen3.8-max",
    api_key = BAILIAN_API_KEY,
    base_url = BAILIAN_API_URL,
)
```

```python
print(model.invoke("简单介绍自己"))
```

#### 1-2 dist字典输入

##### 1-2-1 单轮对话

```python
message = [
    {"role": "system", "content": "你是一个专业的数学老师。"},
    {"role": "user", "content": "请帮我讲解一下勾股定理。"}
]
```

##### 1-2-2 多轮对话 

```python
messages = [
    {"role": "system", "content": "你是一个专业的数学老师。"},
    {"role": "user", "content": "1 + 2 = ？。"},
    {"role": "assistant", "content": "1 + 2 = 3。"},
    { "role": "user", "content": "我刚刚问的是什么问题。"}
]

print(model.invoke(messages))
```

##### 1-2-3 记忆

```python
conversation =[
    {"role": "system", "content": "你是一个专业的数学老师。"},
    {"role": "user", "content": "你叫小斌"}
]

print(model.invoke(conversation).content)

conversation.append({"role": "assistant", "content": model.invoke(conversation).content})
conversation.append({"role": "user", "content": "你叫什么名字？"})

print(f"latest_message: {model.invoke(conversation).content}")
```

#### 1-3 消息列表输入

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage(content="你是一个专业的数学老师。"),
    HumanMessage(content="请帮我简单一句话讲解一下勾股定理。")
]
print(model.invoke(messages).content)

messages = [
    AIMessage(content=model.invoke(messages).content),
    HumanMessage(content="你刚刚讲的是什么内容？")
]
print(f"latest_message: {model.invoke(messages).content}")
```

```python
response = model.invoke("用一句话解释什么是 AI")
# 1. 获取回复内容
print("AI 回复:", response.content)
# 2. 获取响应元数据
metadata = response.response_metadata
print(f"使用的模型: {metadata['model_name']}")
print(f"结束原因: {metadata['finish_reason']}")
print(f"模型提供商：{metadata['model_provider']}\n")
# 3. 获取 Token 使用情况
usage = metadata.get('token_usage', {})
print(f"输入 tokens: {usage.get('prompt_tokens')}")
print(f"输出 tokens: {usage.get('completion_tokens')}")
print(f"总计 tokens: {usage.get('total_tokens')}")
# 4. 获取消息 ID
print(f"消息 ID: {response.id}")
```
