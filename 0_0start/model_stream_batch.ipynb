## 1 流式调用(stream)

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

load_dotenv(override=True)

BAILIAN_API_KEY = os.getenv("BAILIAN_API_KEY")
BAILIAN_API_URL = os.getenv("BAILIAN_API_URL")

model = init_chat_model(
    model = "openai:qwen3.8-max",
    api_key = BAILIAN_API_KEY,
    base_url = BAILIAN_API_URL,
)

for chunk in model.stream("写一首七言律诗，总结大模型的发展"):
    print(chunk.text, end="", flush=True) # 逐token输出
```

## 2 批量调用(batch)

```python
messages = [
    "简单描述中国的地理位置",
    "首都是哪里",
    "中国是的GDP全球排名是多少"
]

batch_resposes = model.batch(messages)
for respone in batch_resposes:
    print(respone.text)
print(batch_resposes)
```

```python
batch_resposes = model.batch_as_completed(messages)
for respone in batch_resposes:
    print(respone)
```

每种调用都有对应的异步方法，函数前加“a”变成异步方法
